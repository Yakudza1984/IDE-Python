```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
sber_data = pd.read_csv('data/sber_data.csv')
sber_data.head()
```

```python
fig = px.box(sber_data, x="ecology", y="price_doc", title="Зависимость цены квартиры от уровня экологии",
             labels={"ecology": "Экологическая обстановка", "price_doc": "Цена квартиры"},
             category_orders={"ecology": sorted(sber_data["ecology"].unique())})  # Сортировка уровней экологии

fig.show()
```

```python
# Строим scatter plot
fig = px.scatter(sber_data, x="kremlin_km", y="price_doc", title="Связь цены квартиры с расстоянием до Кремля",
                 labels={"kremlin_km": "Расстояние до Кремля (км)", "price_doc": "Цена квартиры"},
                 trendline="ols",  # Добавляем линию тренда (линейную регрессию)
                 opacity=0.5)  # Немного прозрачности для точек

fig.show()
```

```python
display(sber_data.isnull().tail())
```

```python
cols_null_percent = sber_data.isnull().mean() * 100
cols_with_null = cols_null_percent[cols_null_percent>0].sort_values(ascending=False)
display(cols_with_null)
```

```python
cols_with_null.plot(
    kind='bar',
    figsize=(10, 4),
    title='Распределение пропусков в данных'
);
```

```python
colors = ['blue', 'yellow'] 
fig = plt.figure(figsize=(10, 4))
cols = cols_with_null.index
ax = sns.heatmap(
    sber_data[cols].isnull(),
    cmap=sns.color_palette(colors),
)
```
